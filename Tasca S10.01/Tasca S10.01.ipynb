## Exploració bàsica amb una API de laboratori

Moltes integracions de negoci depenen de consumir dades d'altres sistemes a través d'APIs (per exemple, dades d'un CRM, d'un ERP, o d'un proveïdor extern). Abans de treballar amb una API real, cal saber com fer peticions bàsiques (consultar dades, crear-ne de noves, modificar-les o eliminar-les) i com interpretar les respostes i els codis d'estat, que indiquen si la petició ha tingut èxit o ha fallat.

En aquest exercici es practicaran alguns dels mètodes HTTP més habituals (GET, POST, PATCH, DELETE) contra una API pública de prova (JSONPlaceholder), simulant les operacions típiques que es farien contra qualsevol servei extern real.

**Definició de la URL base**

Es defineix la URL base de l'API que es farà servir en tot l'exercici.

```python
url_base = "https://jsonplaceholder.typicode.com"
```

**Consulta dels recursos disponibles**

Es realitzen consultes utilitzant el mètode `GET` per obtenir la llista de publicacions, d'usuaris i de tasques.

```python
resposta_posts = requests.get(f"{url_base}/posts")
resposta_users = requests.get(f"{url_base}/users")
resposta_todos = requests.get(f"{url_base}/todos")
```

**Codi d'estat i quantitat de registres**

Es mostra per pantalla el codi d'estat rebut (**200** indica que la petició s'ha completat correctament) i la quantitat total del recurs. Per obtenir aquesta quantitat és necessari convertir la resposta HTTP (que arriba en format JSON) a un objecte Python mitjançant `.json()`, ja que només així es pot aplicar `len()` per comptar els elements. Es mostra per pantalla el codi d’estat rebut (**200** indica que la petició s’ha completat correctament) i la quantitat total del recurs.

```python
print("Posts -> Codi rebut:", resposta_posts.status_code, "| Quantitat:", len(resposta_posts.json()))
print("Users -> Codi rebut:", resposta_users.status_code, "| Quantitat:", len(resposta_users.json()))
print("Todos -> Codi rebut:", resposta_todos.status_code, "| Quantitat:", len(resposta_todos.json()))
```

**Petició a un recurs inexistent**

Es realitza una petició al recurs inexistent "/papallones" i retorna el codi **404**, que indica que el servidor no ha trobat el recurs sol·licitat.

```python
resposta_papallones = requests.get(f"{url_base}/papallones")
print(f"Codi d'estat retornat: {resposta_papallones.status_code}")
```

**Creació d'una nova publicació (POST)**

Per crear una nova publicació, primer es defineix un diccionari amb les dades a enviar (títol, cos del missatge i identificador d'usuari). Es fa la petició `POST` a l'endpoint `/posts`, passant les dades amb el paràmetre `json`. Si la petició té èxit, el servidor retorna un codi **201** (Created) i la publicació creada, incloent un nou `id` assignat automàticament.

```python
dades_enviar_publicacio = {"title": "Puturrú de Fuá", "body": "No te olvides la toalla cuando vayas a la playa", "userId": 1}
resposta_publicacio = requests.post(f"{url_base}/posts", json=dades_enviar_publicacio)
print(f"La resposta JSON és: {resposta_publicacio.json()}")
print(f"El codi d'estat retornat és: {resposta_publicacio.status_code}")
```

**Modificació parcial d'una publicació (PATCH)**

Per modificar parcialment una publicació existent, s'utilitza el mètode `PATCH`, indicant l'`id` del recurs a l'URL i només els camps que es volen canviar. En aquest cas, actualitzem el `title` de la publicació amb `id=1`, mantenint intactes la resta de camps (`body`, `userId`). Un codi **200** (OK) confirma que la modificació s'ha aplicat correctament.

```python
id_publicacio_editar = 1
url_endpoint = f"{url_base}/posts/{id_publicacio_editar}"
dades_enviar_modificacio = {"title": "La cría del berberecho salvaje en la cuenca mediterránea"}
resposta_modificacio = requests.patch(url_endpoint, json=dades_enviar_modificacio)
print(f"Codi d'estat retornat: {resposta_modificacio.status_code}")
print(f"Resposta JSON: {resposta_modificacio.json()}")
```

**Eliminació d'una publicació (DELETE)**

Per eliminar una publicació, s'utilitza el mètode `DELETE`, indicant l'`id` del recurs a l'URL — en aquest cas, la publicació amb `id=1`. Un codi **200** (OK) confirma que la petició s'ha processat correctament. Com que JSONPlaceholder simula l'eliminació sense persistir-la realment, la resposta JSON sol venir buida (`{}`), ja que no hi ha cap contingut a retornar d'un recurs que s'acaba d'eliminar.

```python
id_publicacio_eliminar = 1
url_endpoint = f"{url_base}/posts/{id_publicacio_eliminar}"
resposta_eliminacio = requests.delete(url_endpoint)
print(f"Codi d'estat retornat: {resposta_eliminacio.status_code}")
print(f"Resposta JSON: {resposta_eliminacio.json()}")
```

## Interacció amb una API pública real

Un cop practicats els mètodes bàsics amb una API de laboratori, toca treballar amb una API real i pública, consultant la seva documentació abans d'escriure cap codi. S'ha escollit **Open Trivia Database (OpenTDB)**, una API gratuïta i sense necessitat de clau d'autenticació que retorna preguntes de trivial en format JSON, organitzades per categoria i dificultat.

Es defineix la URL de l'API.

```python
url_trivial = "https://opentdb.com/api.php"
```

**Documentació de l'API**

Consultant la documentació oficial (https://opentdb.com/api_config.php), es troben (entre altres) aquests dos endpoints diferents:

- `GET /api.php?amount=10` — retorna un nombre determinat de preguntes de trivial (per exemple, 10).
- `GET /api_category.php` — retorna el llistat complet de categories disponibles i el seu identificador numèric.

Pel que fa a paràmetres opcionals interessants per filtrar les preguntes:

- `category`: identificador numèric de la categoria (per exemple, 9 = General Knowledge, 21 = Sports).
- `difficulty`: nivell de dificultat (`easy`, `medium` o `hard`).
- `type`: tipus de pregunta (`multiple` per a opció múltiple, `boolean` per a veritat/fals).

La resposta ve en format **JSON**, adequat per treballar-hi directament amb Python, i inclou un camp `response_code` que indica si la petició ha tingut èxit (0) o hi ha hagut algun problema (per exemple, `1` si no hi ha prou preguntes disponibles per als filtres indicats).

**Petició GET senzilla**

Es realitza una petició a l'endpoint principal (`/api.php`), demanant 5 preguntes mitjançant el paràmetre `amount`. Es comprova el codi d'estat de la resposta i, un cop convertida a JSON, s'imprimeix de la categoria, l'enunciat i la resposta correcta de cadascuna de les preguntes rebudes.

```python
resposta_trivia = requests.get(url_trivial, params={"amount": 10})

print(f"Codi d'estat: {resposta_trivia.status_code}")

dades_trivia = resposta_trivia.json()
for pregunta in dades_trivia["results"]:
    print(f"- [{pregunta['category']}] {pregunta['question']} (Resposta: {pregunta['correct_answer']})")
```

**Conversió a DataFrame**

Es converteix la llista de preguntes (`results`) a un DataFrame de pandas, on cada fila representa una pregunta i cada columna un dels seus atributs (categoria, dificultat, enunciat, resposta correcta, respostes incorrectes). Es mostren les primeres files per confirmar que la conversió s'ha fet correctament.

```python
df_trivia = pd.DataFrame(dades_trivia["results"])
df_trivia.head(5)
```

## Interacció amb l'API d'Open Data BCN

L'Ajuntament de Barcelona publica centenars de datasets a través del portal Open Data BCN. En aquest exercici explorarem el catàleg, seleccionarem un dataset amb dades consultables i recuperarem un mínim de 100 registres.

Hem escollit el dataset **"Arbrat de zona de la ciutat de Barcelona"** (`arbrat-zona`), que recull els arbres situats a places, parterres i petits espais enjardinats de la ciutat.

**Cercant un dataset d'interès**

Fem servir el mètode `package_search` per cercar datasets relacionats amb "arbrat". Aquest mètode retorna una llista de datasets que coincideixen amb la cerca, mostrant-ne el títol i l'identificador (`name`), que necessitarem per consultar-ne els detalls.

```python
url_base_bcn = "https://opendata-ajuntament.barcelona.cat/data/api/action"

resposta_cerca = requests.get(f"{url_base_bcn}/package_search", params={"q": "arbrat", "rows": 5})
print(f"Codi d'estat: {resposta_cerca.status_code}")

resultats_cerca = resposta_cerca.json()["result"]["results"]
for dataset in resultats_cerca:
    print(f"- {dataset['title']} (id: {dataset['name']})")
```

**Detall del dataset i selecció del recurs**

Amb `package_show`, obtenim tota la informació del dataset "Arbrat de zona de la ciutat de Barcelona", incloent-hi la llista completa de recursos disponibles. Filtrem els recursos en format CSV o JSON i mostrem el seu `resource_id`, necessari per fer-hi consultes amb `datastore_search`. Seleccionem el recurs CSV que té el `datastore` actiu (`datastore_active: True`), condició imprescindible per poder-lo consultar amb l'API de cerca.

```python
resposta_dataset = requests.get(f"{url_base_bcn}/package_show", params={"id": "arbrat-zona"})
print(f"Codi d'estat: {resposta_dataset.status_code}")

dades_dataset = resposta_dataset.json()["result"]
print(f"Títol: {dades_dataset['title']}")
print(f"Nombre de recursos: {len(dades_dataset['resources'])}")

for recurs in dades_dataset["resources"]:
    if recurs["format"] in ["CSV", "JSON"]:
        print(f"- {recurs['name']} | format: {recurs['format']} | resource_id: {recurs['id']} | datastore actiu: {recurs.get('datastore_active')}")
```

**Recuperació de registres**

Amb el `resource_id` del recurs seleccionat, fem una consulta amb `datastore_search`, indicant un `limit` de 100 registres. La resposta inclou els registres dins del camp `result.records`.

```python
resource_id_arbrat = "29cd5c1f-11b1-404b-b3a5-ae29940b8c55"

resposta_registres = requests.get(f"{url_base_bcn}/datastore_search", params={"resource_id": resource_id_arbrat, "limit": 100})
print(f"Codi d'estat: {resposta_registres.status_code}")

dades_registres = resposta_registres.json()["result"]["records"]
print(f"Registres recuperats: {len(dades_registres)}")
```

**Conversió a DataFrame**

Convertim la llista de registres a un DataFrame de pandas, on cada fila representa un arbre i cada columna un atribut (espècie, adreça, barri, coordenades, etc.).

```python
df_arbrat = pd.DataFrame(dades_registres)
df_arbrat.head()
```

**Desar el resultat i descarregar-lo**

Desem el DataFrame en un fitxer `.csv` dins l'entorn de Colab i, opcionalment, llancem la descàrrega directa a l'ordinador local mitjançant `files.download()`.

```python
df_arbrat.to_csv("arbrat_zona_bcn.csv", index=False)
```
