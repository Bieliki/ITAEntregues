#### obtenir_dades():
1. Demana a l'usuari les dades personals nom, telefon, anys.
2. Demana a l'usuari el nivell respectiu de català, excel i títol acadèmic.
3. Retorna un diccionari amb les dades.

````python
def obtenir_dades():
    """
    Demana a l'usuari les dades per inscriure un candidat.
    Retorna un diccionari amb les dades.
    """

    nom = input("Nom i cognoms: ")
    telefon = input("Telèfon de contacte: ")
    anys = input("Anys d'experiència dins de l'empresa: ")

    print("\nConeixement de català (0–3 punts):")
    print(" 0 = Sense coneixement")
    print(" 1 = Fins a nivell B1")
    print(" 2 = -nivell B2")
    print(" 3 = C1 o superior")
    catala = input("Introdueix un valor (0-3): ")

    print("\nDomini d'Excel (0–3 punts):")
    print(" 0 = Nivell bàsic")
    print(" 1 = Ús de funcions")
    print(" 2 = Taules dinàmiques")
    print(" 3 = Visual Basic i Macros")
    excel = input("Introdueix un valor (0-3): ")

    print("\nTítol acadèmic (0–3 punts):")
    print(" 0 = ESO")
    print(" 1 = FP")
    print(" 2 = Universitat")
    print(" 3 = Màster o superior")
    titol = input("Introdueix un valor (0-3): ")

    return {
        "nom": nom,
        "telefon": telefon,
        "anys": anys,
        "catala": catala,
        "excel": excel,
        "titol": titol
    }
````

#### validar_buits():
1. Recull el diccionari obtingut a obtenir_dades().
2. Retorna False si hi ha algún camp buit.
3. Retorna True si tots els camps tenen algun valor.

````python
def validar_buits(dades):
    """
    Valida que cap camp estigui buit.
    Retorna True si tots els camps tenen algun valor.
    """
    for camp, valor in dades.items():
        if valor == "":
            print(f"Error: el camp '{camp}' no pot estar buit.")
            return False
    return True
````

#### convertir_tipus():
1. Recull el diccionari obtingut a obtenir_dades().
2. Genera una llista amb els camps numèrics.
3. Converteix els valors dels camps numèrics a valors enters.
4. Retorna el diccionari amb els tipus de valors modificats.

````python
def convertir_tipus(dades):
    """
    Converteix els camps numèrics del diccionari a enters.
    Retorna el diccionari modificat si tot és correcte.
    """
    camps_numerics = ["anys", "catala", "excel", "titol"]

    for camp in camps_numerics:
        try:
            dades[camp] = int(dades[camp])
        except ValueError:
            print(f"Error: el camp '{camp}' ha de ser un número vàlid.")
            return False

    return dades
````

#### validar_tipus():
1. Recull el diccionari generat per main_seleccio().
2. Valida que nom i telefon son de tipus str.
3. Retorna True si tot és correcte.

````python
def validar_tipus(dades):
    """
    Valida el tipus de dades introduïdes.
    Retorna True si és correcte.
    """
    esquema = {
        "nom": str,
        "telefon": str
    }

    for camp, tipus_esperat in esquema.items():
        valor = dades.get(camp)
        if not isinstance(valor, tipus_esperat):
            print(f"Error: el camp '{camp}' ha de ser {tipus_esperat.__name__}")
            return False

    return True
````

#### validar_rang():
1. Recull les dades retornades per convertir_tipus().
2. Genera una llista amb els valors dels camps catala, excel i titol.
3. Valida que el valor de cada camp estigui dins del rang indicat. Si està fora del rang envia un missatge d'error i s'atura la validació.
4. Retorna True si tot és correcte.

````python
def validar_rang(dades):
    """
    Valida el rang de les dades catala, excel, titol.
    Retorna True si és correcte.
    """
    camps = ["catala", "excel", "titol"]

    for camp in camps:
        valor = dades.get(camp)
        if not (0 <= valor <= 3):
            print(f"Error: el camp '{camp}' ha de ser un nombre entre 0 i 3")
            return False

    return True
````

#### calcular_puntuacio():
1. Recull els camps anys, catala, excel i titol.
2. Multiplica els anys d'experiència per 1.5.
3. Suma els valors de catala, excel, titol i el resultat de la multiplicació anterior.
4. Retorna la puntuació final (la suma anterior).

````python
def calcular_puntuacio(anys, catala, excel, titol):
    """
    Calcula la puntuació total del candidat.
    Retorna la puntuació total final de cada candidat.
    """
    puntuacio_experiencia = anys * 1.5
    puntuacio_total = puntuacio_experiencia + catala + excel + titol
    return puntuacio_total
````

#### classificar_candidat():
1. Recull la puntuació retornada per calcular_puntuació().
2. Classifica la puntuació a partir dels varems (igual o superior a 12, igual o superior a 8 o menor de 8).
3. Retorna segons la classificació una valoració.

````python
def classificar_candidat(puntuacio):
    """
    Classifica el candidat segons la puntuació total.
    """
    if puntuacio >= 12:
        return "Prioritat Alta"
    elif puntuacio >= 8:
        return "Candidat adequat"
    else:
        return "No prioritari"
````

#### mostrar_resultat():
1. Recull el diccionari generat per main_seleccio().
2. Mostra per pantalla les dades nom, telefon, puntuació i classificació del candidat.

````python
def mostrar_resultat(candidat):
    """
    Mostra per pantalla la informació individual d'un candidat.
    """
    print("\n**** Resultat candidat/da ****")
    print(f"Nom: {candidat['nom']}")
    print(f"Telèfon: {candidat['telefon']}")
    print(f"Puntuació: {candidat['puntuacio']:.2f}")
    print(f"Classificació: {candidat['classificacio']}")
    print("******************************")
````

#### ordenar_prioritat():
1. Recull la classificació obtinguda a classificar_candidat().
2. Ordena la prioritat dels resultats de menor a major.
3. Retorna la prioritat i la puntuació del candidat.

````python
def ordenar_prioritat(candidat):
    """
    Defineix l'ordre de prioritat per a l'informe final.
    """
    prioritat = candidat["classificacio"]

    if prioritat == "Prioritat Alta":
        ordre = 0
    elif prioritat == "Candidat adequat":
        ordre = 1
    else:
        ordre = 2

    return(ordre, -candidat["puntuacio"])
````

#### mostrar_informe():
1. Recull la llista generada a main_seleccio().
2. Ordena la llista per ordre de prioritat.
3. Mostra en pantalla les dades de la llista.

````python
def mostrar_informe(llista_candidats):
    """
    Mostra l'informe final amb tots els candidats ordenats per la puntuació de major a menor.
    """
    print("\n****** INFORME FINAL ******")
    llista_ordenada = sorted(llista_candidats, key=ordenar_prioritat)

    for c in llista_ordenada:
        print(f"Nom: {c['nom']}")
        print(f"Telèfon: {c['telefon']}")
        print(f"Puntuació: {c['puntuacio']:.2f}")
        print(f"Classificació: {c['classificacio']}")
        print("------------------------")

    print("****** FI INFORME ******\n")
````

#### main_seleccio():
1. Genera una llista de candidats.
2. Si les dades introduides son incorrectes envia un missatge per pantalla demanant la re-introducció de les dades.
3. Recull la puntuació retornada per calcular_puntuació.
4. Crida validar_buits().
5. Crida convertir_tipus().
6. Crida validar_tipus().
7. Crida validar_rang().
8. Recull la classificació retornada per classificar_candidat.
9. Genera un diccionari per cada candidat amb el nom, telefon, puntuació i classificació.
10. Afegeix el candidat a la llista de candidats.
11. Executa mostra el resultat per cada candidat.
12. Un cop s'han introduit 10 candidats, informa per pantalla que no es poden introduir més dades de candidats.
13. Mostra l'informe final.


````python
def main_seleccio():
    """
    Ordena tot el procés d'inscripció i classificació.
    """
    candidats = []

    while len(candidats) < 10:
        print(f"\n--- Formulari candidat/a {len(candidats) + 1} ---")

        while True:
            dades = obtenir_dades()

            if not validar_buits(dades):
                print("Si us plau, torna a introduir les dades.\n")
                continue

            dades_convertides = convertir_tipus(dades)
            if not dades_convertides:
                print("Si us plau, torna a introduir les dades.\n")
                continue

            if not validar_tipus(dades_convertides):
                print("Si us plau, torna a introduir les dades.\n")
                continue

            if not validar_rang(dades_convertides):
                print("Si us plau, torna a introduir les dades.\n")
                continue

            break

        puntuacio = calcular_puntuacio(
            dades_convertides["anys"],
            dades_convertides["catala"],
            dades_convertides["excel"],
            dades_convertides["titol"]
        )

        classificacio = classificar_candidat(puntuacio)

        candidat = {
            "nom": dades_convertides["nom"],
            "telefon": dades_convertides["telefon"],
            "puntuacio": puntuacio,
            "classificacio": classificacio
        }

        candidats.append(candidat)

        mostrar_resultat(candidat)

    print("\nJa s'han inscrit 10 persones. No s'accepten més candidats.\n")

    mostrar_informe(candidats)
````

Es generen diccionaris constants per poder realitzar la segmentació.

````python
FACTORS_COMPORTAMENT = {
    0: 0.10,
    1: 0.25,
    2: 0.50,
    3: 0.75,
    4: 0.90
}

FACTORS_EDAT = {
    "18-25": 1.20,
    "26-40": 1.00,
    "41-60": 0.85,
    "60+": 0.70
}

SEGMENTS = {
    0: {"segment": "Navega però no afegeix productes",
        "missatge": "Explora els nostres productes i descobreix noves ofertes per a tu."
    },
    1: {"segment": "Afegeix productes però no arriba al checkout",
        "missatge": "Tens productes a la cistella, torna i revisa la teva selecció."
    },
    2: {"segment": "Arriba al checkout però no paga",
        "missatge": "La teva compra està a punt, finalitza-la en pocs clics."
    },
    3: {"segment": "Cistella de compra abandonat fa menys de 24 h",
        "missatge": "Encara tens la teva cistella disponible, no t'oblidis d'emportar-te el que més t'agrada."
    },
    4: {"segment": "Cistella de compra abandonat fa més de 24 h",
        "missatge": "Recupera la teva cistella amb una oferta especial pensada per a tu."
    }
}
````

#### obtenir_segment_factor():
1. Recupera el codi_comportament de Clients.
2. Recupera el segment i missatge corresponent de SEGMENTS.
3. Recupera el factor corresponent de FACTOR_COMPORTAMENT.
4. Retorna una tupla amb el segment, el missatge i el factor.

````python
def obtenir_segment_factor(client):
    """
    Va a buscar en els diccionaris constants SEGMENTS i FACTORS_COMPORTAMENT.
    Retorna una tupla amb el segment, missatge i factor_comportament.
    """
    codi = client["codi_comportament"]
    info = SEGMENTS[codi]
    factor = FACTORS_COMPORTAMENT[codi]
    return info["segment"], info["missatge"], factor
````

#### obtenir_factor_edat():
1. Recupera el grup_edat de clients.
2. Retorna el factor_edat corresponent de FACTORS_EDAT.

````python
def obtenir_factor_edat(client):
    """
    Va a buscar el grup_edat de la llista de diccionaris clients.
    Retorna el factor_edat corresponent.
    """
    grup = client["grup_edat"]
    return FACTORS_EDAT[grup]
````

#### calcular_risc():
1. Rep el factor_comportament i el factor_edat de processar_clients().
2. Retorna el producte dels dos factors.

````python
def calcular_risc(factor_comportament, factor_edat):
    """
    Rep factor_comportament i factor_edat.
    Retorna el producte de les dues dades.
    """
    return factor_comportament * factor_edat
````

#### processar_clients():
1. Rep llista de diccionaris _clients_.
2. Genera la llista clients_processats.
3. Recull segment, missatge i factor_comportament de obtenir_segment_factor().
4. Recull factor_edat de obtenir_factor_edat().
5. Recull risc_final de calcular_risc().
6. Genera el diccionari nou_client amb les dades del client i les dades rebudes.
7. Afegeix aquest diccionari a la llista clients_processats.
8. Retorna la llista clients_processats.

````python
def processar_clients(clients):
    """
    Recull segment, missatge, factor_comportament, factor_edat i risc_final.
    Retorna llista clients_processats.
    """
    clients_processats = []

    for client in clients:
        segment, missatge, factor_comportament = obtenir_segment_factor(client)
        factor_edat = obtenir_factor_edat(client)
        risc_final = calcular_risc(factor_comportament, factor_edat)

        nou_client = {
            "id_client": client["id_client"],
            "nom": client["nom"],
            "edat": client["edat"],
            "grup_edat": client["grup_edat"],
            "codi_comportament": client["codi_comportament"],
            "segment": segment,
            "missatge": missatge,
            "factor_comportament": factor_comportament,
            "factor_edat": factor_edat,
            "risc_final": risc_final
        }

        clients_processats.append(nou_client)

    return clients_processats
````

1. Rep les dades de processar_clients_seguro().
2. Calcula el preu de la pòlissa a partir d'una formula estipulada anteriorment.

````python
def calcular_preu_polissa(preu_base, edat):
    """
    Rep dades de preu i edat.
    Retorna cálcul del preu de la pòlissa.
    """
    return preu_base * (1 + edat/100)
````

#### recopilar_caracters():
1. Rep els paràmetres booleans de main_contrasenya().
2. Recopila les dades corresponents als paràmetres que son True.
3. Retorna diccionari.

````python
def recopilar_caracters(majuscules, minuscules, numeros, signes):
    """
    Rep els paràmetres booleans.
    Retorna diccionari.
    """
    conjunts = {}

    if majuscules:
        conjunts["majuscules"] = list(string.ascii_uppercase)
    if minuscules:
        conjunts["minuscules"] = list(string.ascii_lowercase)
    if numeros:
        conjunts["numeros"] = list(string.digits)
    if signes:
        conjunts["signes"] = list(string.punctuation)

    return conjunts
````

#### definir_minims():
1. Rep el generador aleatori i el diccionari conjunts.
2. Genera la llista minims.
3. Selecciona un caracter aleatori de cada conjunt.
4. Afegeix el caràcter a la llista minims.
5. Retorna la llista minims.

````python
def definir_minims(aleatori, conjunts):
    """
    Rep el generador aleatori i un diccionari.
    Retorna llista.
    """
    minims = []
    for llista in conjunts.values():
        minims.append(aleatori.choice(llista))
    return minims
````

#### definir_restants():
1. Rep el generador aleatori, el diccionari conjunts i la quantitat de caràcters restants.
2. Genera la llista tots.
3. Omple la llista amb tots els caràcters disponibles.
4. Selecciona aleatoriaments tants caràcters com la quantitat de caràcters restants rebuda.
5. Retorna la llista amb els caràcters restants seleccionats.

````python
def definir_restants(aleatori, conjunts, longitud_rest):
    """
    Rep el generador aleatori, un diccionari i una quantitat.
    Retorna llista.
    """
    tots = []
    for llista in conjunts.values():
        tots.extend(llista)
    return list(aleatori.choice(tots, size=longitud_rest))
````

#### remenar_caracters():
1. Rep el generador aleatori i una llista amb els caracters (minims + restants).
2. Desordena els caràcters de la llista.
3. Retorna la llista sense espais i convertida a string.

````python
def remenar_caracters(aleatori, llista):
    """
    Rep el generador aleatori i una llista.
    Retorna llista convertida a string sense espais.
    """
    aleatori.shuffle(llista)
    return "".join(llista)
````

#### main_contrasenya():
1. Rep els paràmetres longitud, majuscules, minuscules, numeros i signes.
2. Crea el generador aleatori amb la funció np.random.default_rng().
3. Crida la funció recopilar_caracters() i recupera diccionari conjunts.
4. Comproba si conjunts està buit i en aquest cas mostra avís per pantalla.
5. Crida la funció definir_minims() i recupera la llista minims.
6. Crida la funció definir_restants() i recupera llista restants.
7. Genera la llista contrasenya concatenant les llistes minims i restants.
8. Crida la funció remenar_caracters() i rep un string amb la contrasenya final.
9. Retorna la contrasenya.

````python
def main_contrasenya(longitud=12, majuscules=True, minuscules=True, numeros=True, signes=False):
    """
    Ordena tot el procés per generar contrasenya aleatòria.
    Retorna contrasenya aleatòria.
    """

    aleatori = np.random.default_rng()

    conjunts = recopilar_caracters(majuscules, minuscules, numeros, signes)

    if not conjunts:
        raise ValueError("Has de seleccionar com a mínim un tipus de caràcter.")

    minims = definir_minims(aleatori, conjunts)

    restants = definir_restants(aleatori, conjunts, longitud - len(minims))

    contrasenya = minims + restants

    return remenar_caracters(aleatori, contrasenya)
````

#### netejar_linia():
1. Rep una linia equivalent a un partit.
2. Treu els espais sobrants.
3. Comproba que no estigui buida.
4. Ho separa en paraules.
5. Retorna la linia.

````python
def netejar_linia(linia):
    """
    Rep una linia i la neteja.
    Retorna linia netejada.
    """
    linia = linia.strip()
    if not linia:
        return None
    return linia.split()
````

#### desglossar_partit():
1. Rep les paraules netes.
2. Busca la posicio del marcador.
3. Separa el nom de l'equip local i visitant.
4. Extreu els gols del marcador.
5. Retorna un diccionari.

````python
def desglossar_partit(parts):
    """
    Rep les paraules netes.
    Retorna un diccionari.
    """
    posicio_marcador = next(posicio for posicio, paraula in enumerate(parts) if "-" in paraula)

    equip_local = " ".join(parts[:posicio_marcador])
    marcador = parts[posicio_marcador]
    equip_visitant = " ".join(parts[posicio_marcador+1:])

    gols_local, gols_visitant = map(int, marcador.split("-"))

    return {
        "local": equip_local,
        "visitant": equip_visitant,
        "gols_local": gols_local,
        "gols_visitant": gols_visitant
    }
````

#### llegir_partits():
1. Rep el nom del fitxer a llegir.
2. Crea la llista partits.
3. Envía cada linia a netejar_linia().
4. Envía la linia neta a desglossar_partit().
5. Afegeix els partits a la llista.
6. Retorna la llista partits.

````python
def llegir_partits(nom_fitxer):
    """
    Rep el nom del fitxer a llegir.
    Retorna la llista partits.
    """
    partits = []

    with open(nom_fitxer, "r", encoding="utf-8") as f:
        for linia in f:
            parts = netejar_linia(linia)
            if parts is None:
                continue

            partit = desglossar_partit(parts)
            partits.append(partit)

    return partits
````

#### construir_dataframe():
1. Rep la llista partits.
2. Converteix la llista al dataframe df_futbol.
3. Retorna el dataframe df_futbol.

````python
def construir_dataframe(partits):
    """
    Rep la llista partits.
    Retorna el dataframe df_futbol.
    """
    df_futbol = pd.DataFrame(partits)
    return df_futbol
````

#### crear_equip():
1. Rep el diccionari equips i un nom.
2. Comprova si l'equip existeix en el diccionari.
3. Si no existeix, crea l'equip al diccionari amb les estadístiques a 0.

````python
def crear_equip(equips, nom):
    """
    Rep el diccionari equips i un nom.
    Crea l'equip al diccionari amb les estadístiques a 0.
    """
    if nom not in equips:
        equips[nom] = {
            "gf": 0,
            "gc": 0,
            "guanyats": 0,
            "empatats": 0,
            "perduts": 0,
            "punts": 0
        }
````

#### actualitzar_gols():
1. Rep el diccionari equips i les dades del partit.
2. Suma els gols a favor i en contra de l'equip local.
3. Suma els gols a favor i en contra de l'equip visitant.

````python
def actualitzar_gols(equips, local, visitant, gols_local, gols_visitant):
    """
    Rep el diccionari equips i les dades del partit.
    Suma els gols a favor i en contra de l'equip local.
    Suma els gols a favor i en contra de l'equip visitant.
    """
    equips[local]["gf"] += gols_local
    equips[local]["gc"] += gols_visitant
    equips[visitant]["gf"] += gols_visitant
    equips[visitant]["gc"] += gols_local
````

#### actualitzar_resultat():
1. Rep el diccionari equips i les dades del partit.
2. Determina si ha guanyat l'equip local, el visitant o si ha estat empat.
3. Assigna els punts corresponents.

````python
def actualitzar_resultat(equips, local, visitant, gols_local, gols_visitant):
    """
    Rep el diccionari equips i les dades del partit.
    Assigna els punts corresponents per victoria, derrota o empat.
    """
    if gols_local > gols_visitant:
        equips[local]["guanyats"] += 1
        equips[visitant]["perduts"] += 1
        equips[local]["punts"] += 3

    elif gols_visitant > gols_local:
        equips[visitant]["guanyats"] += 1
        equips[local]["perduts"] += 1
        equips[visitant]["punts"] += 3

    else:
        equips[local]["empatats"] += 1
        equips[visitant]["empatats"] += 1
        equips[local]["punts"] += 1
        equips[visitant]["punts"] += 1
````

#### processar_partits():
1. Rep el diccionari equips i les dades del partit.
2. Crida crear_equip() per crear els equips.
3. Crida actualitzar_gols() per actualitzar els gols.
4. Crida actualitzar_resultat() per actualitzar els resultats.

````python
def processar_partit(equips, local, visitant, gols_local, gols_visitant):
    """
    Rep el diccionari equips i les dades del partit.
    Crea els equips si cal i actualitza gols i resultats.
    """
    crear_equip(equips, local)
    crear_equip(equips, visitant)
    actualitzar_gols(equips, local, visitant, gols_local, gols_visitant)
    actualitzar_resultat(equips, local, visitant, gols_local, gols_visitant)
````

#### actualitzar_equips():
1. Rep el dataframe df_futbol.
2. Crea el diccionari equips.
3. Crida processar_partit() per procesar cada linia (partit).
4. Retorna el diccionari equips.

````python
def actualitzar_equips(df_futbol):
    """
    Rep el dataframe df_futbol.
    Retorna el diccionari equips.
    """
    equips = {}

    for _, fila in df_futbol.iterrows():
        processar_partit(equips, fila["local"], fila["visitant"], fila["gols_local"], fila["gols_visitant"])

    return equips
````

#### generar_classificacio():
1. Rep el diccionari equips.
2. Converteix el diccionari a dataframe df_futbol.
3. Ordena el dataframe df_futbol per punts, gols a favor i gols en contra.
4. Retorna el dataframe df_futbol.



````python
def generar_classificacio(equips):
    """
    Rep el diccionari equips.
    Retorna el dataframe df_futbol ordenat.
    """
    df_futbol = pd.DataFrame.from_dict(equips, orient="index")

    df_futbol = df_futbol[["punts", "gf", "gc", "guanyats", "empatats", "perduts"]]

    df_futbol = df_futbol.sort_values(
        by=["punts", "gf", "gc"],
        ascending=[False, False, True]
    )

    return df_futbol
````
